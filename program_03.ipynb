### Implement Gradient Descent Function

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def gradient_descent(x, y):

    # my algorithem - may not be correct
    # initially take m and b as 0
    # calculate the error (mse)
    # calculate the slope for m in the plane m and mse and slope of b in the plane b and mse (basically calculate the derivative of the mse with respect m and b)
    # take small step towards the slope - calculate new values of m and b using the formula x = x - learningrate * d/dx
    # repeate until the mse of the previous is smaller than the new one

    m_curr, b_curr = 0, 0
    iterations = 1000
    n = len(x)
    lr = 0.08 # lerning rate

    # remember here x and y are not single values - they are numpy arrays

    plt.figure(figsize=(15, 6))
    for i in range(iterations):
        y_pred = m_curr * x + b_curr # all predicted y values
        mse = (sum([val ** 2 for val in (y - y_pred)])) / n
        md = -(2/n) * sum(x * (y - y_pred)) # d(mse)/dm
        bd = -(2/n) * sum(y - y_pred) # d(mse)/db


        m_curr = m_curr - (lr * md)
        b_curr = b_curr - (lr * bd)

        plt.plot(x, m_curr * x + b_curr, color="blue" if i == iterations-1 else "cyan")
        plt.title("Gradient Descent")
        plt.ylabel("y")
        plt.xlabel("x")

        print("m: {}, b: {}, cost: {}, iteration: {}".format(m_curr, b_curr, mse, i))
        
    plt.scatter(x, y, color="red", linewidths=2, zorder=3)
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 7, 9, 11, 13])

# expected result -> m = 2 and b = 3

gradient_descent(x, y)
```
