### Simple Linear Regression

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv("./data/simple_house_prices.csv")
df.head()
```

```python
plt.figure(figsize=(11, 6))
plt.scatter(df.size_sqft, df.price_inr, marker='+', color="red")
plt.xlabel("House Size (sqft)")
plt.ylabel("House price (inr)")
```

```python
lr_model = LinearRegression()
lr_model.fit(df[["size_sqft"]], df["price_inr"])
```

```python
print(lr_model.predict([[1000]]))
```

```python
m = lr_model.coef_  # slope m
c = lr_model.intercept_  # y intercetp c

plt.figure(figsize=(11, 6))
plt.scatter(df.size_sqft, df.price_inr, marker="+", color="red")
plt.plot(df.size_sqft, m*df.size_sqft + c, color="blue")
plt.xlabel("House Size (sqft)")
plt.ylabel("House price (inr)")

```
