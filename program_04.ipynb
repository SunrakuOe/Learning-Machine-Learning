### Practice Gradient Descent

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("./data/practice_gradient_descent_with_student_marks.csv")
x, y = df.math, df.cs
```

```python
def gradient_descent(x, y, lr = 0.001, iterations = 10000):
    m = b = 0
    learning_rate = lr
    iterations = iterations
    n = len(x)

    for _ in range(iterations):
        y_pred = m * x + b
        md = -(2/n) * sum(x * (y - y_pred)) # d(mse)/dm
        bd = -(2/n) * sum(y - y_pred) # d(mse)/db

        m = m - learning_rate * md
        b = b - learning_rate * bd
        # print("m: {}, b: {}, cost: {}".format(m, b, sum([val * val for val in y - (m * x + b)]) / n))

    return m, b,  sum([val * val for val in y - (m * x + b)]) / n
```

```python
m, b, cost = gradient_descent(x, y, 0.00009, 100000)
print("m: {}, b: {}, cost: {}".format(m, b, cost))
```

```python
plt.scatter(x, y, color="red", marker="+")
plt.plot(x, m * x + b, color="blue")
```
