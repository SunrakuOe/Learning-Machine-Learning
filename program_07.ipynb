# Training and Testing Data

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("./data/car_data.csv")
df.head()
```

```python
fig, (axes1, axes2) = plt.subplots(1, 2, figsize=(12, 4))
axes1.scatter(df.milage, df.sell_price_inr, marker="+", color="red")
axes1.set_title("milage vs sell price")
axes1.set_xlabel("milage")
axes1.set_ylabel("sell price (inr)")
axes2.scatter(df.age_yr, df.sell_price_inr, marker="+", color="red")
axes2.set_title("age(year) vs sell price")
axes2.set_xlabel("age(year)")
axes2.set_ylabel("sell price (inr)")
```

```python
train_x, test_x, train_y, test_y = train_test_split(df.drop(columns=["sell_price_inr"]), df.sell_price_inr, test_size=0.2)
```

```python
model = LinearRegression().fit(train_x, train_y)
```

```python
model.predict(test_x)
model.score(test_x, test_y)
```
