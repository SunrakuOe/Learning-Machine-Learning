## Lgistic regression (multiclass classification)
- assignment of the class do do logicatic regression on sklearn's iris dataset

```python
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sn
import matplotlib.pyplot as plt
```

```python
iris = load_iris()
print(dir(iris))
print(len(iris.data))
```

```python
for i in range(5):
    print("data: {}, target: {}, targe_names: {}".format(iris.data[i], iris.target[i], iris.target_names[iris.target[i]]))

iris.target_names
```

```python
x_train, x_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2)
```

```python
model = LogisticRegression().fit(x_train, y_train)
```

```python
print(model.predict([iris.data[0]]))
print(model.predict([iris.data[50]]))
```

```python
y_pred = model.predict(x_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
plt.figure(figsize=(6,5))
sn.heatmap(cm, annot=True)
plt.xlabel("Truth")
plt.ylabel("Predicted")
```
