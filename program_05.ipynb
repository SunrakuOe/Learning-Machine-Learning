### Saving a model using pickle module

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report
import pickle
```

```python
df = pd.read_csv("./data/simple_house_prices.csv")
df.head()
```

```python
plt.figure(figsize=(11, 6))
plt.scatter(df.size_sqft, df.price_inr, marker='+', color="red")
plt.xlabel("House Size (sqft)")
plt.ylabel("House price (inr)")
```

```python
lr_model = LinearRegression()
lr_model.fit(df[["size_sqft"]], df["price_inr"])
```

```python
print(lr_model.predict([[1000]]))
```

```python
with open("./model/house_price_pickle", "wb") as f:
    pickle.dump(lr_model, f)
```

```python
with open("./model/house_price_pickle", "rb") as f:
    retrived_model = pickle.load(f)
```

```python
print(retrived_model.predict([[1000]]))
```

### Save a model using joblib

```python
import joblib
```

```python
joblib.dump(lr_model, "./model/house_price_model_joblib")
```

```python
retrived_model = joblib.load("./model/house_price_model_joblib")
```

```python
print(retrived_model.predict([[1000]]))
```
